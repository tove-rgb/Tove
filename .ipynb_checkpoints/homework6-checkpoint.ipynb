```python
import numpy as np
```

```python
list = [20, 27, 9, 12, 31, 17, 4, 50, 39, 35]
arr = np.array(list)

print(arr * 4)
print(np.mean(arr))
print(np.median(arr))
print(np.std(arr))
indexmin = np.argmin(arr)
indexmax = np.argmax(arr)
print(arr[indexmin])
print(arr[indexmax])
print(np.min(arr))
print(np.max(arr))
# TF and how to do boolean indexing
print(arr[arr>20])
```
