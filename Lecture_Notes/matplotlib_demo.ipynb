```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# sin function
x = np.linspace (0, 2*np.pi , 100) #creates points 100 points in between 0 and 2pi
y = np.sin(x)
```

```python
plt.figure(figsize = (6,4))
plt.plot(x,y)
plt.show()
```

```python
y1 = np.sin(x)
y2 = np.cos(x)
```

```python
plt.figure()

plt.scatter(x, y1, label = "sin(x)")
plt.scatter(x, y2, label = "cos(x)", color = "cyan")

plt.title("since and cosine wave")
plt.xlabel("x values (radians)")
plt.ylabel("f(x)")
plt. legend()

plt.show
```

```python
# Customization
plt.figure()
plt.plot(x, y1, color = "xkcd:icky green", linestyle = "-", marker = "*", alpha = 0.3)

plt.show()
```

```python
# Subplots 

fig, axes = plt.subplots(1, 2, figsize = (10,4))

axes[0].plot(x,y1, color = "teal", ls = ":")

axes[0].set_title("Sine Wave")

axes[1].plot(x, y2, color = "cornflowerblue", ls = "--")

axes[1].set_title("Cosine Wave", color = "darkblue")

plt.show
```

```python
x = np.linspace(0,2*np.pi,200)
fig, axes = plt.subplots(1,3, figsize=(12, 4))

axes[0].plot(x, np.sin(x)) 
axes[0].set_title("Function of Sin(x)")
axes[0].set_xlabel("x")
axes[0].set_ylabel("sin(x)")

axes[1].plot(x, np.sin(2 * x)) 
axes[1].set_title("Function of Sin(2x)")
axes[1].set_xlabel("x")
axes[1].set_ylabel("sin(2x)")

axes[2].plot(x, np.sin(3*x)) 
axes[2].set_title("Function of Sin(3x)")
axes[2].set_xlabel("x")
axes[2].set_ylabel("sin(3x)")

plt.show()
```

```python
x = np.linspace(0 , 10, 100)

a_vals = [0,0.5,1,1.5,2]

plt.figure()

for a in a_vals:
    plt.plot(x , x**a , label = f"x^{a}")
    
plt.legend()
plt. title ("Function od x^a")
plt.xlabel("x")
plt.ylabel("x^a")
plt.show()
```

```python
def arb_plotter(func_name):
    current_func = getattr(np, func_name)
    x = np.linspace(-10,10,400)
    y = current_func(x)
    
    plt.figure()
    plt.plot(x,y)
    plt.title(f"Function Name of {func_name}(x)")
    plt.xlabel("x")
    plt.ylabel(f"{func_name}(x)")
    plt.show()

arb_plotter("arctan")
```
