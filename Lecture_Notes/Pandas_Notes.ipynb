```python
data.values()
```

```python
data["Names"]
```

```python
df = pd.DataFrame(data)
df
```

```python
# Access Column
df['Names']
```

```python
# Access Multiple Columns
df[['Names','City']]
```

```python
# Filter
df[df['Ages']>30]
```

```python
df[df['City'] == 'Berkeley']
```
